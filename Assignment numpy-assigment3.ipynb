# Question 2

```python
arr = np.random.randint(1, 101, 10) #array of 10 values from 1 to 100
x = int(input("enter a value X: "))
diffs = np.abs(arr - x) #calculate the distance or the difference
min_diff = np.min(diffs)                
closest_values = np.unique(arr[diffs == min_diff]) #keeps only closest values and removes repeated values
closest_list = list(closest_values) #convert the numpy array to a normal list
output_str = ""
for i, val in enumerate(closest_list):
    if i == 0:
        output_str = str(val)  # First item
    else:
        output_str += " or " + str(val)  

if len(closest_list) > 1:
    print(f"{output_str} (whichever is mathematically closer)")
else:
    print(f"{output_str}")
```

# Question 5

```python
arr = np.random.randint(1, 51, 30)
arr = np.where((arr % 3 == 0) | (arr % 5 == 0), -1, arr)
print(arr)
```
